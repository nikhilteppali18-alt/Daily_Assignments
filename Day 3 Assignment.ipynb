1.Write a function to check whether a string is palindrome or not.

```python
n=input("Enter the String")
def p():
  if n==n[::-1]:
    print("Palidrome",n)
  else:
    print("Not Palidrome",n)
p()
```

2.Write a function to find the second largest number in a list.

```python
def second_largest(numbers):
    largest = second = float('-inf')

    for num in numbers:
        if num > largest:
            second = largest
            largest = num
        elif num > second and num != largest:
            second = num

    return second


nums = [10, 20, 4, 45, 99]
print(second_largest(nums))
```

3.Write a function that accepts a list and returns only unique values.

```python
def unique_values(my_list):
    return list(set(my_list))
numbers = [1, 2, 2, 3, 4, 4, 5]
print(unique_values(numbers))
```

4.Use lambda to find the square of a number.

```python
square = lambda x: x * x
print(square(5))
```

5.Use lambda with map() to convert a list of integers into squares.

```python
numbers = [1, 2, 3, 4, 5]

squares = list(map(lambda x: x * x, numbers))

print(squares)
```

6.Use lambda with filter() to extract even numbers from a list.

```python
numbers = [1, 2, 3, 4, 5, 6, 7, 8]

even_numbers = list(filter(lambda x: x % 2 == 0, numbers))

print(even_numbers)
```

7.Use lambda with sorted() to sort a list of tuples based on second value.

```python
students = [("John", 85), ("Alice", 92), ("Bob", 78)]

sorted_list = sorted(students, key=lambda x: x[1])

print(sorted_list)
```

8.Use lambda with reduce() to find the product of all numbers in a list.

```python
from functools import reduce

numbers = [1, 2, 3, 4, 5]

product = reduce(lambda x, y: x * y, numbers)

print(product)
```

9.Use lambda to find the maximum between two numbers.

```python
maximum = lambda a, b: a if a > b else b

print(maximum(10, 25))
```

10.Create Calculator using Functions

Operations:

Add

Subtract

Multiply

Divide

Use lambda for operations.

```python
add = lambda a, b: a + b
subtract = lambda a, b: a - b
multiply = lambda a, b: a * b
divide = lambda a, b: a / b if b != 0 else "Cannot divide by zero"
def calculator():
    print("Calculator Operations")
    print("1. Add")
    print("2. Subtract")
    print("3. Multiply")
    print("4. Divide")

    choice = int(input("Enter your choice (1-4): "))

    num1 = float(input("Enter first number: "))
    num2 = float(input("Enter second number: "))

    if choice == 1:
        print("Result:", add(num1, num2))

    elif choice == 2:
        print("Result:", subtract(num1, num2))

    elif choice == 3:
        print("Result:", multiply(num1, num2))

    elif choice == 4:
        print("Result:", divide(num1, num2))

    else:
        print("Invalid Choice")

calculator()
```
