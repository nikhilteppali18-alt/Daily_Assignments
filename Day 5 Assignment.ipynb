7. Create the following dataframe

| Student | Marks | City      |
| ------- | ----- | --------- |
| A       | 85    | Hyderabad |
| B       | NaN   | Chennai   |
| C       | 92    | NaN       |
| D       | 78    | Bangalore |
| E       | NaN   | Mumbai    |

```python
import numpy as np
df1=pd.DataFrame({
    "Student":['A','B','C','D','E'],
    "Marks":[85,np.nan,92,78,np.nan],
    "City":["Hyderbad","Chennai",np.nan,"Bangalore","Mumbai"]

})
```

```python
df1
```

 7.1 Find total missing values in each column.

```python
df1.isnull().sum()
```

7.2 Display rows containing missing values.

```python
print("Row containing null values:")
df1[df1.isnull().any(axis=1)]
```

7.3 Replace missing marks with average marks.

```python
df1['Marks'] = df1['Marks'].fillna(df1['Marks'].mean())
df1
```

7.4 Replace missing city values with "Unknown".


```python
df1['City'] = df1['City'].fillna("Unknown")
df1
```

7.5 Drop rows having any missing value.

```python
df1
```

 7.6 Drop rows only if Marks is missing.

```python
df1.dropna(inplace=True)
```

```python
df1
```

```python
df1.dropna(subset=['Marks'], inplace=True)
```

```python
df1
```

9. Create two Dataframes

| Emp_ID | Name  |
| ------ | ----- |
| 101    | Ravi  |
| 102    | Priya |
| 103    | Kiran |
| 104    | Sneha |

```python
df3=pd.DataFrame({
    "Emp_ID":[101,102,103,104],
    "Name":["Rahul","Priya","Kiran","Sneha"]
})
```

```python
df3
```


| Emp_ID | Department |
| ------ | ---------- |
| 101    | IT         |
| 102    | HR         |
| 103    | Finance    |
| 105    | Marketing  |

```python
df4=pd.DataFrame({
    "Emp_ID":[101,102,103,105],
    "Department":["IT","HR","Finance","Marketing"]

})
```

```python
df4
```

9.1 Perform Inner Join.

```python
pd.merge(df3, df4, how='inner')
```

 9.2 Perform Left Join.

```python
pd.merge(df3, df4, how='left')
```

9.3 Perform Right Join.

```python
pd.merge(df3, df4, how='right')
```

9.4 Perform Outer Join.

```python
pd.merge(df3, df4, how='outer')
```

9.5 Identify employees present in one DataFrame but not the other.

```python
result = pd.merge(
    df3,
    df4,
    on='Emp_ID',
    how='outer',
    indicator=True
)

print(result[result['_merge'] != 'both'])
```
